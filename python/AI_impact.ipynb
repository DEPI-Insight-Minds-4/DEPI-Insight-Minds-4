```python
# تثبيت المكتبات (مرة واحدة فقط)
!pip install plotly ipywidgets

# الاستيراد
import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display

# قراءة البيانات
project = pd.read_csv(r'C:\Users\O.R\OneDrive\Desktop\ai_job_trends_dataset.csv')

# قائمة الصناعات المتاحة
industries = ['All'] + sorted(project['Industry'].dropna().unique().tolist())

# عنصر اختيار الصناعة
industry_dropdown = widgets.Dropdown(
    options=industries,
    value='All',
    description='الصناعة:',
    style={'description_width': 'initial'}
)

# دالة لتحديث الرسومات
def update_dashboard(selected_industry):
    if selected_industry == 'All':
        data = project
    else:
        data = project[project['Industry'] == selected_industry]

    # الرسم 1: العلاقة بين نسبة الإناث ومخاطر الأتمتة
    fig1 = px.scatter(
        data,
        x='Gender Diversity (%)',
        y='Automation Risk (%)',
        color='Industry',
        title='العلاقة بين نسبة الإناث ومخاطر الأتمتة',
        hover_name='Industry',
        size='Automation Risk (%)'
    )

    # الرسم 2: الصناعات الأعلى في العمل عن بُعد
    remote_avg = (
        project.groupby('Industry')['Remote Work Ratio (%)']
        .mean()
        .sort_values(ascending=False)
        .head(10)
        .reset_index()
    )
    fig2 = px.bar(
        remote_avg,
        x='Industry',
        y='Remote Work Ratio (%)',
        title='أعلى 10 صناعات من حيث العمل عن بُعد',
        color='Remote Work Ratio (%)'
    )

    # الرسم 3: معدل مخاطر الأتمتة حسب الصناعة
    risk_avg = (
        project.groupby('Industry')['Automation Risk (%)']
        .mean()
        .sort_values(ascending=False)
        .reset_index()
    )
    fig3 = px.bar(
        risk_avg,
        x='Industry',
        y='Automation Risk (%)',
        title='معدل مخاطر الأتمتة حسب الصناعة',
        color='Automation Risk (%)'
    )

    # عرض الرسومات
    fig1.show()
    fig2.show()
    fig3.show()


# تفعيل التفاعل
widgets.interact(update_dashboard, selected_industry=industry_dropdown)
```
