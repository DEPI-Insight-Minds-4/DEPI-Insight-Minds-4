```python

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, Markdown


file_path = r'C:\Users\O.R\OneDrive\Desktop\ai_job_trends_dataset.csv'
project = pd.read_csv(file_path, encoding='utf-8')

# تحويل الأعمدة الرقمية
num_cols = ['Median Salary (USD)', 'Job Openings (2024)', 'Projected Openings (2030)', 
            'Remote Work Ratio (%)', 'Automation Risk (%)', 'Gender Diversity (%)']
for col in num_cols:
    project[col] = pd.to_numeric(project[col], errors='coerce')


avg_salary = project['Median Salary (USD)'].mean()
avg_auto_risk = project['Automation Risk (%)'].mean()
avg_remote = project['Remote Work Ratio (%)'].mean()
avg_gender = project['Gender Diversity (%)'].mean()
growth_jobs = project['Projected Openings (2030)'].sum() - project['Job Openings (2024)'].sum()

display(Markdown(f"""
# 🧠 **AI Job Trends – Analytical Dashboard**
---
**💰 متوسط الراتب:** `${avg_salary:,.0f}`  
**🤖 متوسط خطر الأتمتة:** `{avg_auto_risk:.1f}%`  
**🌐 متوسط نسبة العمل عن بُعد:** `{avg_remote:.1f}%`  
**👩‍💼 متوسط التنوع بين الجنسين:** `{avg_gender:.1f}%`  
**📈 نمو فرص العمل المتوقعة:** `{growth_jobs:,.0f}`
"""))


fig1 = px.scatter(
    project,
    x='Automation Risk (%)',
    y='Median Salary (USD)',
    color='Industry',
    size='Job Openings (2024)',
    hover_name='Industry',
    title='💸 العلاقة بين خطر الأتمتة ومتوسط الراتب حسب الصناعة',
    trendline='ols'
)
fig1.show()

# =======================================================
# 4️⃣ العلاقة بين العمل عن بُعد وخطر الأتمتة
# =======================================================
fig2 = px.scatter(
    project,
    x='Remote Work Ratio (%)',
    y='Automation Risk (%)',
    color='AI Impact Level',
    size='Median Salary (USD)',
    title='🌐 العلاقة بين العمل عن بُعد وخطر الأتمتة',
    hover_name='Industry'
)
fig2.show()


corr = project[num_cols].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("📊 خريطة الترابط بين المؤشرات")
plt.show()


ai_jobs = project.groupby('AI Impact Level')[['Job Openings (2024)', 'Projected Openings (2030)']].sum().reset_index()

fig3 = px.bar(
    ai_jobs,
    x='AI Impact Level',
    y=['Job Openings (2024)', 'Projected Openings (2030)'],
    barmode='group',
    text_auto='.2s',  # 🟢 عرض الأرقام فوق الأعمدة
    title='📈 تأثير الذكاء الاصطناعي على تطور فرص العمل',
    labels={'value': 'عدد الفرص', 'AI Impact Level': 'مستوى تأثير الذكاء الاصطناعي'}
)
fig3.update_traces(textposition='outside')  # 🟢 إظهار النص فوق العمود
fig3.update_layout(xaxis_tickangle=0, bargap=0.25)
fig3.show()

# =======================================================
# 7️⃣ تحليل الصناعات الأفضل من حيث الأداء العام (محسّن)
# =======================================================
project['Score'] = (
    project['Median Salary (USD)'].rank(pct=True) * 0.4 +
    project['Gender Diversity (%)'].rank(pct=True) * 0.3 -
    project['Automation Risk (%)'].rank(pct=True) * 0.3
)

top_industries = project.groupby('Industry')['Score'].mean().sort_values(ascending=False).head(10).reset_index()

fig4 = px.bar(
    top_industries,
    x='Industry',
    y='Score',
    text='Score',  # 🟢 عرض القيمة فوق العمود
    title='🏆 أفضل الصناعات من حيث الراتب، التنوع، وقلة خطر الأتمتة',
    labels={'Industry': 'الصناعة', 'Score': 'النتيجة المركبة'}
)
fig4.update_traces(texttemplate='%{text:.2f}', textposition='outside', marker_color='#64B5F6')
fig4.update_layout(xaxis_tickangle=35, bargap=0.4)
fig4.show()


fig5 = px.box(
    project,
    x='AI Impact Level',
    y='Gender Diversity (%)',
    color='AI Impact Level',
    title='👩‍💻 توزيع التنوع بين الجنسين حسب تأثير الذكاء الاصطناعي'
)
fig5.show()
```
